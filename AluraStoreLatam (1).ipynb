### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
print(tienda.columns)
print(tienda2.columns)
print(tienda3.columns)
print(tienda4.columns)
```

```python
facturacion_tienda_1 = tienda["Precio"].sum()
facturacion_tienda_2 = tienda2["Precio"].sum()
facturacion_tienda_3 = tienda3["Precio"].sum()
facturacion_tienda_4 = tienda4["Precio"].sum()

print("Facturación total por tienda:")
print(f"Tienda 1: ${facturacion_tienda_1:,.2f}")
print(f"Tienda 2: ${facturacion_tienda_2:,.2f}")
print(f"Tienda 3: ${facturacion_tienda_3:,.2f}")
print(f"Tienda 4: ${facturacion_tienda_4:,.2f}")
```

```python
import matplotlib.pyplot as plt

nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
facturaciones = [facturacion_tienda_1, facturacion_tienda_2, facturacion_tienda_3, facturacion_tienda_4]

plt.figure(figsize=(8,5))
plt.bar(nombres_tiendas, facturaciones, color='skyblue')
plt.title('Facturación Total por Tienda')
plt.ylabel('Monto en $')
plt.xlabel('Tiendas')
plt.tight_layout()
plt.show()
```

# 2. Ventas por categoría

```python
categorias_tienda_1 = tienda['Categoría del Producto'].value_counts()
print("Categorías más vendidas - Tienda 1:")
print(categorias_tienda_1)
categorias_tienda_2 = tienda2['Categoría del Producto'].value_counts()
print("Categorías más vendidas - Tienda 2:")
print(categorias_tienda_2)
categorias_tienda_3 = tienda3['Categoría del Producto'].value_counts()
print("Categorías más vendidas - Tienda 3:")
print(categorias_tienda_3)
categorias_tienda_4 = tienda4['Categoría del Producto'].value_counts()
print("Categorías más vendidas - Tienda 4:")
print(categorias_tienda_4)
```

```python
import matplotlib.pyplot as plt
import numpy as np

facturacion_tienda_1 = tienda["Precio"].sum()
facturacion_tienda_2 = tienda2["Precio"].sum()
facturacion_tienda_3 = tienda3["Precio"].sum()
facturacion_tienda_4 = tienda4["Precio"].sum()

fig, ax1 = plt.subplots(figsize=(12, 6))

nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
facturaciones = [facturacion_tienda_1, facturacion_tienda_2, facturacion_tienda_3, facturacion_tienda_4]
ax1.bar(nombres_tiendas, facturaciones, color='skyblue', label='Facturación')

ax1.set_xlabel("Tiendas")
ax1.set_ylabel("Facturación Total ($)", color='skyblue')
ax1.tick_params(axis='y', labelcolor='skyblue')

ax2 = ax1.twinx()

cat1 = tienda['Categoría del Producto'].value_counts().head(5)
cat2 = tienda2['Categoría del Producto'].value_counts().head(5)
cat3 = tienda3['Categoría del Producto'].value_counts().head(5)
cat4 = tienda4['Categoría del Producto'].value_counts().head(5)

categorias = set(cat1.index) | set(cat2.index) | set(cat3.index) | set(cat4.index)
categorias = sorted(categorias)

valores1 = [cat1.get(cat, 0) for cat in categorias]
valores2 = [cat2.get(cat, 0) for cat in categorias]
valores3 = [cat3.get(cat, 0) for cat in categorias]
valores4 = [cat4.get(cat, 0) for cat in categorias]

x = np.arange(len(categorias))
width = 0.2

ax2.bar(x - 1.5*width, valores1, width, label='Tienda 1 (Categorías)', alpha=0.5)
ax2.bar(x - 0.5*width, valores2, width, label='Tienda 2 (Categorías)', alpha=0.5)
ax2.bar(x + 0.5*width, valores3, width, label='Tienda 3 (Categorías)', alpha=0.5)
ax2.bar(x + 1.5*width, valores4, width, label='Tienda 4 (Categorías)', alpha=0.5)

ax2.set_ylabel("Cantidad de ventas por categoría", color='orange')
ax2.tick_params(axis='y', labelcolor='orange')

plt.xticks(x, categorias, rotation=45)
fig.tight_layout()

ax2.legend(loc="upper left", bbox_to_anchor=(1, 1), title="Categorías Vendidas")

plt.title("Facturación Total y Categorías Más Vendidas por Tienda")
plt.show()
```

# 3. Calificación promedio de la tienda


```python
calif_prom_tienda_1 = tienda['Calificación'].mean()
calif_prom_tienda_2 = tienda2['Calificación'].mean()
calif_prom_tienda_3 = tienda3['Calificación'].mean()
calif_prom_tienda_4 = tienda4['Calificación'].mean()

print("Calificación promedio por tienda:")
print(f"Tienda 1: {calif_prom_tienda_1:.2f}")
print(f"Tienda 2: {calif_prom_tienda_2:.2f}")
print(f"Tienda 3: {calif_prom_tienda_3:.2f}")
print(f"Tienda 4: {calif_prom_tienda_4:.2f}")

import matplotlib.pyplot as plt

nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
calificaciones = [calif_prom_tienda_1, calif_prom_tienda_2, calif_prom_tienda_3, calif_prom_tienda_4]

plt.figure(figsize=(8,5))
plt.bar(nombres_tiendas, calificaciones, color='mediumseagreen')
plt.title('Calificación Promedio por Tienda')
plt.ylabel('Calificación (de 1 a 5)')
plt.ylim(0, 5)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

# 6 Analisis Final


El Sr. Juan está considerando vender sus productos en una de las cuatro tiendas disponibles en Alura Store. Para ayudarle a tomar una decisión informada y estratégica, se llevó a cabo un análisis detallado de los datos de ventas de cada tienda, considerando:

*  Ingresos totales por tienda

*  Categorías de productos más populares

*  Productos más y menos vendidos

*  Costo promedio de envío hacia los clientes

El objetivo es identificar cuál tienda ofrece el mejor entorno de ventas, en términos de rentabilidad, alcance y satisfacción del cliente.

**Análisis**

1. Ingresos Totales por Tienda
Se calcularon los ingresos totales sumando los valores de los productos vendidos. El resultado mostró diferencias notables entre tiendas. En un gráfico de barras, se visualizó que:

* Tienda 2 lidera en facturación, con ventas significativamente mayores.

* Tienda 3 también muestra buenos ingresos.

* Tiendas 1 y 4 tienen ingresos menores, aunque competitivos.


---



2. Categorías Más Populares
Cada tienda tiene su propia especialización. El análisis de frecuencia reveló:

* Tienda 2 tiene una gran variedad, pero destaca en electrónica y accesorios.

* Tienda 3 se enfoca en productos del hogar.

* Tienda 1 tiene una demanda repartida, sin una categoría dominante clara.

* Tienda 4 vende mayormente artículos de oficina y escolares.

Esto indica que el perfil del producto del Sr. Juan debe alinearse con la tienda seleccionada.


---

3. Calificaciones Promedio de Clientes
Se calculó la calificación promedio de cada tienda (de 1 a 5). Los resultados fueron:

* Tienda 3 obtuvo la mejor calificación promedio (~4.67), lo que refleja alta satisfacción del cliente.

* Tienda 2 también recibió calificaciones positivas (~4.45).

* Tiendas 1 y 4 tuvieron puntajes más bajos (~4.12 y ~4.00 respectivamente).

La atención al cliente y la calidad del servicio parecen estar mejor en tiendas 2 y 3.

---

4. Productos Más y Menos Vendidos
Se identificaron los productos con mayor y menor rotación en cada tienda. Esto reveló:

* Tienda 2 tiene productos estrella con alta rotación, lo que puede ser ideal para ventas rápidas.

* Tienda 3 mantiene un portafolio más equilibrado, pero con menos variabilidad.

* Tienda 1 tiene menos dinamismo en ventas.

* Tienda 4 mostró alta dispersión, sin un claro líder de ventas.

---

Finalmente, se analizó el costo promedio de envío por tienda:

* Tienda 3 presentó el costo de envío más bajo (~$8.20).

* Tienda 2 tuvo un costo moderado (~$9.15).

* Tiendas 1 y 4 mostraron los costos más altos (~$10.30 y ~$11.05).

Esto influye directamente en la competitividad final del producto para el cliente.




---

**Conclusión y Recomendación**
Recomendación: El Sr. Juan debería vender sus productos en la Tienda 2.

**Justificación:**

*  Es la tienda con mayor facturación, lo que demuestra un alto volumen de ventas.

* Posee categorías con gran demanda, especialmente en productos electrónicos, lo cual puede ser compatible con los productos del Sr. Juan.

* Tiene buenas calificaciones de clientes, lo cual refleja confianza y satisfacción.

* Cuenta con productos que se venden bien y con rapidez, ideal para una estrategia de rotación.

* Aunque no tiene el costo de envío más bajo, este es razonable y no representa una desventaja significativa frente a su alto rendimiento.

Tienda 3 sería una segunda opción sólida, especialmente si el Sr. Juan busca un enfoque en satisfacción del cliente y costos logísticos bajos, aunque con menos volumen de ventas.

 De igual manera, este análisis debe revisarse periódicamente, ya que el comportamiento de los consumidores, las tendencias y la competencia pueden cambiar. También se recomienda que el Sr. Juan evalúe el tipo de producto que desea vender para asegurar el mejor encaje con la tienda seleccionada.
